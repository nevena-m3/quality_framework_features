```python
from pathlib import Path
import sys

def find_project_root(start):
    start = Path(start).resolve()
    for candidate in (start, *start.parents):
        if (candidate / 'config' / 'project.yaml').is_file() and (candidate / 'src').is_dir():
            return candidate
    raise FileNotFoundError('Could not locate the paper_1 project root.')

ROOT = find_project_root(Path.cwd())
for source_root in (ROOT / 'src reviewed', ROOT / 'src'):
    if str(source_root) not in sys.path:
        sys.path.insert(0, str(source_root))

from paper1_qc_reviewed.qdist_v410_freeze_readiness import finalize_measurement
print(f'Project root: {ROOT}')
```
