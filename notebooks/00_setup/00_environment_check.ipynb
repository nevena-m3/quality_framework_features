# 00 — Environment and configuration check

Checks the local configuration and executable entry point before reading participant data.

Core algorithms live in `src/paper1_qc`; this notebook is an auditable orchestration/reporting layer.

```python
from pathlib import Path
import subprocess, sys
import pandas as pd

def find_project_root():
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / 'pyproject.toml').exists():
            return candidate
    raise FileNotFoundError('Run this notebook from inside the paper_1 project.')

ROOT = find_project_root()
CONFIG = ROOT / 'config' / 'project.yaml'
OUTPUT = ROOT / 'outputs'
print('Project:', ROOT)
print('Config:', CONFIG)
```

```python
assert CONFIG.exists(), 'Copy config/project.example.yaml to config/project.yaml and review it.'
subprocess.run([sys.executable, '-m', 'paper1_qc.cli', '--help'], cwd=ROOT, check=True)
```
