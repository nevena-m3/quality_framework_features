# 02 — Window, support, and sensitivity audit

This notebook addresses the short-window concern directly.

A short local window is acceptable only when the phenomenon is local and the
recording-level estimate is supported by enough independent evidence. The audit
therefore examines availability, support fields, and the presence of existing
window/parameter/deletion/invariance evidence.


```python
from pathlib import Path
import pandas as pd

from paper1_qc.external_validation import (
    discover_project_root,
    support_audit,
)

PROJECT_ROOT = discover_project_root()
paths = support_audit(PROJECT_ROOT)
paths
```

```python
support = pd.read_csv(paths["support_summary"])
evidence = pd.read_csv(paths["evidence_inventory"])

display(
    support.sort_values(
        ["family_code", "role", "missing_fraction"],
        ascending=[True, True, False],
    )
)
print("Sensitivity evidence files discovered:", len(evidence))
display(evidence.head(100))
```

## Evidence that must be signed feature by feature

For each indicator, determine whether the reviewed outputs already establish:

- plausible window/hop/guard stability;
- leave-one-pause, boundary, segment, or event stability;
- minimum and typical evidence counts;
- segmentation or boundary sensitivity;
- resampling/invariance behavior;
- known-truth dose response;
- competing-mechanism discrimination.

The automated inventory only locates evidence. It does not decide whether the
evidence is scientifically sufficient.


```python
# Create a local review table without overwriting any frozen family artifact.
review = support.copy()
review["window_parameter_stability"] = "PENDING"
review["leave_one_unit_out_stability"] = "PENDING"
review["boundary_segmentation_sensitivity"] = "PENDING"
review["effective_evidence_adequacy"] = "PENDING"
review["reviewer_note"] = ""

review_path = (
    PROJECT_ROOT
    / "outputs"
    / "06_external_validation"
    / "02_window_support"
    / "window_support_manual_review.csv"
)
if not review_path.exists():
    review.to_csv(review_path, index=False)
print(review_path)
```
