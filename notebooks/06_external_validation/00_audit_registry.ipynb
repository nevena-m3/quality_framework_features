# 00 — Freeze the external-validation audit registry

This notebook creates the feature-level audit ledger and freezes the
indicator–comparator hypotheses **before** examining results.

External measures are classified as equivalent, related primitives,
algorithmic comparators, perceptual comparators, or complementary measures.
They are not treated as ground truth.


```python
from pathlib import Path
import pandas as pd

from paper1_qc.external_validation import (
    AUDIT_VERSION,
    discover_project_root,
    load_latest_release,
    write_audit_registry,
)

PROJECT_ROOT = discover_project_root()
PROJECT_ROOT
```

```python
features, registry = load_latest_release(PROJECT_ROOT)

print(f"Audit version: {AUDIT_VERSION}")
print(f"Recordings in latest reviewed release: {len(features):,}")
print(f"Reviewed indicators: {len(registry):,}")

display(
    registry[
        [
            "family_code",
            "feature",
            "role",
            "unit",
            "worse",
            "default_analysis",
        ]
    ].sort_values(["family_code", "feature"])
)
```

```python
paths = write_audit_registry(PROJECT_ROOT)
paths
```

```python
audit_registry = pd.read_csv(paths["indicator_registry"])
comparators = pd.read_csv(paths["comparator_registry"])

display(audit_registry)
display(comparators)
```
