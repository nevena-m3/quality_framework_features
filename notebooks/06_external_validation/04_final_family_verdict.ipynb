## Final decision principle

External tools can reveal convergence, blind spots, and failure modes. They do
not determine validity by popularity. The final verdict must integrate the
mathematical implementation, known-truth behavior, local-window stability,
recording-level support, competing mechanisms, human case review, and the
narrow claim the feature is intended to support.
